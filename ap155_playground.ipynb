```python
import numpy as np
import matplotlib.pyplot as plt

n_steps = 1000       # Total iterations per r value
mark = 900     # Number of iterations to plot (after transients die out)
r_resolution = 10000 # How many r values to simulate between 2.5 and 4.0

r = np.linspace(1.0, 4.0, r_resolution)

x = np.ones(r_resolution)*rng.random()

plt.figure(figsize=(12, 8), dpi=300)

for i in range(n_steps):
    x = r * x * (1 - x) # The logistic map function
    
    if i >= mark:
        plt.plot(r, x, ',k')

plt.ylim(0, 1)
plt.grid(True, alpha=0.3)
plt.show()
```
