### python3 venv

- `venv` is Python's built-in module for creating isolated virtual environments.
- It provides each project with its own self-contained directory for Python packages, keeping its dependencies completely separated from your computer's global Python installation and other projects.

(macOS/Linux)
1. Create the Virtual Environment: `python3 -m venv .venv`
2. Activate virtual environment: `source .venv/bin/activate`
3. Deactivate virtual environment: `deactivate`

- **BEST PRACTICE:** when working on a project especially when installing dependencies, always do so within the virtual environment

### Problem 3

- The logistic map is a polynomial recurrence relation that models how populations change over discrete generations
  $$ x_{n+1} = r x_n (1-x_n)$$
  where $x_n$ is the population at year $n$ ($x\in[0,1]$), $x_{n+1}$ is the population in the following year, and $r$ is the growth rate ($r \in [1,4]$).
- A bifurcation diagram is a plot of the fixed points (x^*) on the (y)-axis as a function of the growth rate (r) on the (x)-axis. A fixed point is a value of (x) that remains unchanged under the recursion relation, meaning that once the system reaches (x^*), subsequent iterations remain at (x^*).
- An interesting visualization is obtained: ![Figure](figures/ap155_0_bifurcationdiagram.png)

#### (Hint) algorithm:
1. Import necessary libraries (`numpy` and `matplotlib.pyplot`)
2. Generate 1000 $r$ values from 1.0 to 4.0.
3. Generate 1000 random values for $x_0$ ranging between 0.0 to 1.0.
4. For each $x_0$ and $r$, generate 1000 values of $x_n$.
5. We only want the fixed points, so we discard all transient values. From the 1000 values generated for each $x_0$-$r$ pair, only retain the last 200 values.
6. For each $r$ (x-axis), plot all 200 corresponding $x^*$.
