- Objective: Familiarize with Python coding fundamentals (dealing with `numpy` and `matplotlib`)
- Problem: Plot the bifurcation diagram of the logistic growth
      - The logistic map is a polynomial recurrence relation that models how populations change over discrete generations
      $$ x_{n+1} = r x_n (1-x_n)$$ where $x_n$ is the population at year $n$ ($x\in[0,1]$), $x_{n+1}$ is the population in the following year, and $r$ is the growth rate ($r \in [1,4]$).
      - A bifurcation diagram

- Algorithm:

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

n_steps = 1000       
mark = 900     # Number of iterations to plot (after transients die out)
r_resolution = 10000 # How many r values to simulate between 2.5 and 4.0

r = np.linspace(1.0, 4.0, r_resolution)

x = np.ones(r_resolution)*rng.random()

plt.figure(figsize=(12, 8), dpi=300)

for i in range(n_steps):
    x = r * x * (1 - x) # The logistic map function
    
    if i >= mark:
        plt.plot(r, x, ',k')

plt.ylim(0, 1)
plt.xlabel(r'Growth rate (r)')
plt.xlim(2.5, 4.0)
plt.ylabel(r'Fixed point (x^{*})')
plt.grid(True, alpha=0.3)
plt.savefig('figures/ap155_0_bifurcationdiagram.pdf', format='pdf')
plt.show()
```
