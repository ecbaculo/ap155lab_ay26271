## Setting Up SSH and Git

Git is a version control system that allows you to keep track of changes to your files. In this course, you will use Git to save and version your code, notebooks, and figures on the server, and synchronize them with a private GitHub repository.

The basic workflow will eventually be:

JupyterLab/Server --> Git --> GitHub --> Technical Report

### Create a GitHub Account

1. Go to GitHub.
2. Create an account using an email address that you can access.
3. Verify your email address.
4. Remember your GitHub username and the email address associated with your account.
5. You do not need to remember a GitHub password for server authentication if you will use an SSH key.

- **[Important] Never share your GitHub password or SSH private key with anyone.**

### Open the Server Terminal
1. Log in to the course server and open a Terminal (You will use this terminal to run Git commands)

### Check That Git Is Installed
1. In the terminal, enter: `git --version`
2. You should see something similar to: `git version 2.34.1`
3. You can also check where Git is installed: `which git` (which will show `/usr/bin/git`)
4. If you see a Git version and a path to Git, Git is available on the server.

- **[Important] If git: command not found appears, do not try to install Git yourself. Contact the instructor or server administrator.**

### Check current directory
1. To see your current location on the server, enter: `pwd` (example: `/home/jovyan/work/ap155`)
2. To see the files and folders in your current directory: `ls`
3. For a more detailed listing, you can use: `ls -la`

### Navigating between directories
- To enter a folder: `cd <folder-name>` (For example: `cd ap155`)
- To move up one directory: `cd ..`
- To return to your home directory: `cd ~`
- **Tip: You can use the Tab key to autocomplete file and folder names in the terminal.**

### Check if current directory is already a Git repo
1. Navigate to your assigned project directory, then run: `git status`
2. If the directory is already a Git repository, you should see something similar to: `On branch main ...` (This means Git is already tracking your project.)
3. If you see: `fatal: not a git repository` (Do not immediately run git init.) Your course repository may already be configured elsewhere. Ask your instructor where you should work.

### Check Whether an SSH Key Already Exists
SSH keys allow the server to securely authenticate with GitHub without requiring you to enter your GitHub password every time.
1. Check whether an SSH directory already exists: `ls -la ~/.ssh`
2. You may see files such as: `id_ed25519,  id_ed25519.pub` (The files come in pairs: id_ed25519 is the private key, id_ed25519 is the public key)
3. The private key must remain private.
4. Never:
   - send your private key to anyone;
   - upload it to GitHub;
   - put it in your project repository;
   - paste it into a group chat;
   - include it in a report or screenshot.
5. The .pub file is the public key and may be added to your GitHub account when instructed.
6. If an SSH key already exists, do not overwrite it without first asking the instructor/server administrator.

### Generate an SSH Key if You Do Not Already Have One
1. If your instructor confirms that you need a new SSH key and you do not already have one, run: `ssh-keygen -t ed25519 -C "your-github-email@up.edu.ph"`
2. Replace the email with the email associated with your GitHub account.
3. When asked where to save the key, press Enter to use the default location: `/home/jovyan/.ssh/id_ed25519`
4. You will also be asked for a passphrase. (enter your password, remember this)
5. After successful creation, you should have:
    `~/.ssh/id_ed25519`
    `~/.ssh/id_ed25519.pub`

### Add Your Public SSH Key to GitHub
1. Display your public key: `cat ~/.ssh/id_ed25519.pub`
2. You should see one long line beginning with: `ssh-ed25519 ...`
3. Copy the entire line.
4. Then, on GitHub: Open Settings > Select SSH and GPG keys  > Click New SSH key > Give the key a title and **paste your public key into the key field**
5. Save the key.
** [IMPORTANT] Only copy the contents of `id_ed25519.pub`. Never copy `id_ed25519`.**

### Test Your GitHub SSH Connection
1. Return to the server terminal and run: `ssh -T git@github.com`
2. The first time you connect, you may be asked whether you trust GitHub's host key.
3. If everything is configured correctly, GitHub should respond with a message indicating that you have successfully authenticated.
    - You may see something similar to: `Hi <your-username>! You've successfully authenticated, but GitHub does not provide shell access.`
    - This is a successful result.

### Important Git Commands
Once your GitHub repository is connected to your server, you will commonly use:
1. Check the current status:  `git status`
2. Add a file to the next commit: `git add <filename>`
3. Save your changes as a commit: `git commit -m "Describe your changes"`
4. Upload your commits to GitHub: `git push`
5. Download changes from GitHub: `git pull`

### Cloning
1. Create a public repository named `git-practice` on GitHub.
2. Clone it to your local machine and enter the directory: `git clone <url>`
3. Create a file named `analysis.py` containing a function to calculate the mean of a given list.
4. Save, stage, commit, and push:
    - `git add analysis.py`
    - `git commit -m "Add mean calculation"`
    - `git push origin main`
5. Create and switch to a feature branch: `git switch -c add-median`
6. Update `analysis.py` to include a median calculator
7. Commit and push the branch to GitHub:
    - `git add analysis.py`
    - `git commit -m "Add median calculation"`
    - `git push -u origin add-median`
8. Go to your repository on GitHub.
9. Open a Pull Request to merge add-median into main.
10. Review the code diff, click Create Pull Request, and then click Merge Pull Request.
11. Switch back to main locally and pull the newly merged code:
    - `git switch main`
    - `git pull origin main`
12. Intentionally introduce a bug in analysis.py by changing the mean function
13. Commit the bad change (do not push yet):
    - `git commit -am "Modify mean calculation"`
14. Check your commit history to copy the hash of the bad commit (e.g., a1b2c3d): `git log --oneline`
15. Revert only that commit and push to GitHub
    - `git revert <bad-commit-hash>`
    - `git push origin main`
16. Create a README.md file explaining your repository layout and commit history.
17. Push your final documentation:
    - `git add README.md`
    - `git commit -m "Add workflow documentation"`
    - `git push origin main`
